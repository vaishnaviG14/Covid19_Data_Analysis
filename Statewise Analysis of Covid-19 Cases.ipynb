# STATEWISE ANALYSIS OF COVID CASES

```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as snb
import numpy as np

df = pd.read_csv (r'state_level_daily1.csv')   #read the csv file (put 'r' before the path string to address any special characters in the path, such as '\'). Don't forget to put the file name at the end of the path + ".cs
```

## Bar plot of confirmed cases of each state

```python
w=0.9
st_name = df["State_Name"]
con_cases = df["Confirmed"]
color = ['r', 'g', 'b', 'y', 'k']
plt.figure(figsize=(10,7))
plt.barh(st_name.iloc[1:39] , con_cases[1:39] , w, color=color);
```

```python
plt.figure(figsize=(17,8))
st_name=df['State'].iloc[1:39]
con_cases=df['Confirmed'].iloc[1:39]
snb.scatterplot(st_name, con_cases, data=df, sizes=20, hue='State_Name', s=100)
plt.legend(bbox_to_anchor=(1, 1.2), loc='upper left', borderaxespad=0)
```

## Bar chart recovered cases of each state

```python
w=0.9
st_name = df["State_Name"]
rv_cases = df["Recovered"]
color = ['r', 'g', 'b', 'y', 'k']
plt.figure(figsize=(10,7))
plt.barh(st_name.iloc[1:39] , rv_cases[1:39] , w, color=color);
```

## Confirmed VS deceased Scatterplot

```python
plt.figure(figsize=(17,8))
con_cases=df['Confirmed'].iloc[1:39]
dtt=df['Deceased'].iloc[1:39]
snb.scatterplot(con_cases, dtt, data=df, s=100, hue='State_Name')
plt.legend(bbox_to_anchor=(1, 1.2), loc='upper left', borderaxespad=0)
```

## Confirmed and Recovered cases of each state 

```python
plt.figure(figsize=(17,8))
st_name = df["State"][1:39]
rv_cases = df["Recovered"][1:39]
con_cases = df["Confirmed"][1:39]
plt.scatter(st_name, con_cases, color='red', label='Confirmed')
plt.scatter(st_name, rv_cases, color='blue', label='Recovered')
plt.legend()
```

## Confirmed and deceased cases of each state

```python
plt.figure(figsize=(17,8))
st_name = df["State"][1:39]
con_cases = df["Confirmed"][1:39]
dtt = df["Deceased"][1:39]

plt.plot(st_name, con_cases, color='g', label='Confirmed')
plt.plot(st_name, dtt, color='r', label='Deceased')
plt.legend()
```

## Pie Chart overview of total confirmed, deceased and recovered cases

```python
plt.figure(figsize=(17,8))
con_total = df["Confirmed"][1:39].sum()
rv_total = df["Recovered"][1:39].sum()
dtt_total = df["Deceased"][1:39].sum()
total = (con_total, rv_total, dtt_total)

cmap = plt.get_cmap('Spectral')
colors = [cmap(i) for i in np.linspace(0, 1, 8)]

source_pie = plt.pie(total, labels=["Confirmed", "Recovered", "Deceased"], autopct='%1.1f%%', shadow=True, colors=colors)
```

## Pie chart of first 10 states reported largest number of confirmed cases on 1st april 2020

```python
df1 = pd.read_csv (r'C:\Users\Hp\Desktop\state_level_daily2.csv')
plt.figure(figsize=(10, 10))
cmap = plt.get_cmap('Spectral')
colors = [cmap(i) for i in np.linspace(0, 1, 8)]

df1.drop(df1[df1['State'] == 'TT'].index, inplace = True)
date1 = df1.head(38)
sorted_df = date1.sort_values(by=["Confirmed"], ascending=False)

st_label = sorted_df["State_Name"][0:10]
con_total = sorted_df["Confirmed"][0:10]

#source_pie = plt.pie(con_total, labels=st_label, shadow=True, colors=colors)
plt.pie(con_total, colors=colors, labels= con_total, counterclock=False, shadow=True)
plt.legend(st_label, bbox_to_anchor=(1, 1.2), loc='upper left', borderaxespad=0)
```
